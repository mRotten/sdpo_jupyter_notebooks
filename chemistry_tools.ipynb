```python
from rdkit import Chem
from rdkit.Chem import Draw
from pathlib import Path

def smiles_to_image(smiles, img_name, parent_folder):
    """
    Generates and saves a 2-dimensional chemical structure image from a SMILES string.

    Parameters:
        smiles (str): The SMILES string of the molecule.
        img_name (str): The filename to save the image.
    """
    mol = Chem.MolFromSmiles(smiles)
    if mol:
        Draw.MolToFile(mol, parent_folder / img_name, size=(1000, 1000))
        print(f"Image saved as {img_name}")
    else:
        print("Invalid SMILES string.")
```

```python
# Example usage
parent_folder = Path("C:/Users/mlalonde/Documents/jupyter_sandbox/chemical_structures")
file_name = 'molecule.png'
smiles_string = "CC1=CC(C2=CN(CC3CCC(F)(F)CC3)N=C2OCC2CCN(CCC3=CC=C4C(C5CCC(=O)NC5=O)=NN(C)C4=C3)CC2)=NC(OC2=C3C(=CC=C2)N(C)C(=O)N3)=C1"
smiles_to_image(smiles_string, file_name, parent_folder)
```
