```python
import requests
from bs4 import BeautifulSoup

def get_html_from_assay_id(assay_id):
    url = f"https://www.eurofinsdiscovery.com/catalog/{assay_id}"
    headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}

    try:
        response = requests.get(url, headers=headers)
        response.raise_for_status()
    except requests.RequestException as e:
        print(f"Error fetching assay page {assay_id}: {e}")
        return None
    return response.text

def sandbox(response):
    soup = BeautifulSoup(response.text, 'html.parser')

    # Look for UniProt ID in the page text or table (adjust selector as needed)
    text = soup.get_text()
    for line in text.splitlines():
        if "UniProt" in line:
            parts = line.split()
            for part in parts:
                if part.startswith("P") and len(part) == 6:  # Typical UniProt ID pattern
                    return part

    print(f"UniProt ID not found for assay {assay_id}")
    return None

# Example usage
assay_ids = [2338, 2387, 3001]
assay_pages = list()

for assay_id in assay_ids:
    assay_pages.append(get_html_from_assay_id(assay_id))

print(assay_pages[0])
```
